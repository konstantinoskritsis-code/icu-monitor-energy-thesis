## 1. MIMIC-IV Exploration
### File: icustays.csv.gz
This file contains one row per ICU stay with admission and discharge timestamps.
It is the foundation of the MIMIC-IV analysis.

```python
# Load MIMIC-IV ICU stays
df_icustays = pd.read_csv('/Users/dinoskritsis/Desktop/MIMIC/icustays.csv.gz', 
                           compression='gzip')

print("=== MIMIC-IV ICU Stays ===")
print("Total ICU stays:", df_icustays['stay_id'].nunique())
print("Total unique patients:", df_icustays['subject_id'].nunique())
print("\nFirst 5 rows:")
df_icustays.head()
```

```python
# ICU unit type breakdown
print("=== ICU Unit Types — MIMIC-IV ===")
print(df_icustays['first_careunit'].value_counts())
```

```python
# Length of stay statistics
print("=== Length of Stay — MIMIC-IV (days) ===")
print(df_icustays['los'].describe().round(2))
```

```python
# Missing data check
print("=== Missing Data — MIMIC-IV icustays ===")
print(df_icustays.isnull().sum())
```

### File: chartevents.csv.gz

## 2. eICU Collaborative Research Database Exploration

The eICU database is particularly significant for this project because it was 
collected directly through the Philips eICU telehealth program. The vitalPeriodic 
table contains data recorded automatically by Philips IntelliVue bedside monitors 
at 5-minute intervals — making it the most direct evidence of Philips monitor 
activity available in any public dataset.

```python
# Load eICU patient file
df_patient = pd.read_csv('/Users/dinoskritsis/Desktop/E-ICU/patient.csv.gz',
                          compression='gzip')

print("=== eICU Patient File ===")
print("Total ICU stays:", df_patient['patientunitstayid'].nunique())
print("Total unique patients:", df_patient['uniquepid'].nunique())
print("\nFirst 5 rows:")
df_patient.head()
```

```python
# ICU unit types and length of stay
print("=== ICU Unit Types — eICU ===")
print(df_patient['unittype'].value_counts())

print("\n=== Length of Stay — eICU (hours) ===")
df_patient['los_hours'] = df_patient['unitdischargeoffset'] / 60
print(df_patient['los_hours'].describe().round(2))
```

```python
# Missing data check
print("=== Missing Data — eICU patient file ===")
print(df_patient.isnull().sum())
```

### File: vitalPeriodic.csv.gz
This file contains vital signs recorded automatically by Philips bedside monitors 
at 5-minute intervals with no human involvement.
Every row represents the monitor actively running and transmitting data.
Full file contains 146,671,642 rows — one of the largest public clinical datasets.

```python
# Load vitalPeriodic — sample for structure exploration
df_periodic = pd.read_csv('/Users/dinoskritsis/Desktop/E-ICU/vitalPeriodic.csv.gz',
                           compression='gzip', nrows=1000)

print("=== vitalPeriodic — eICU (1000 row sample) ===")
print("Shape:", df_periodic.shape)
print("\nColumns:", df_periodic.columns.tolist())
print("\nFirst 5 rows:")
df_periodic.head()
```

```python
# Full missing data assessment on complete vitalPeriodic file
# Warning — this takes a few minutes to load
print("Loading full vitalPeriodic file — please wait...")

df_periodic_full = pd.read_csv('/Users/dinoskritsis/Desktop/E-ICU/vitalPeriodic.csv.gz',
                                compression='gzip')

print("=== Full Missing Data Assessment — vitalPeriodic ===")
print("Shape:", df_periodic_full.shape)
print("\nMissing values:")
print(df_periodic_full.isnull().sum())
```

### File: vitalAperiodic.csv.gz
Captures vital sign measurements that do not occur on a fixed schedule —
primarily non-invasive blood pressure readings at irregular intervals.
Together with vitalPeriodic it provides a complete picture of monitoring activity.
Full file contains 25,075,074 rows.

```python
# Full missing data assessment on complete vitalAperiodic file
print("Loading full vitalAperiodic file — please wait...")

df_vital_full = pd.read_csv('/Users/dinoskritsis/Desktop/E-ICU/vitalAperiodic.csv.gz',
                              compression='gzip')

print("=== Full Missing Data Assessment — vitalAperiodic ===")
print("Shape:", df_vital_full.shape)
print("\nMissing values:")
print(df_vital_full.isnull().sum())
```

## 3. Key Findings Summary
A combined overview of all exploration findings across both datasets.

```python
# Combined dataset scale summary
print("=" * 50)
print("COMBINED DATASET SCALE")
print("=" * 50)

mimic_stays = df_icustays['stay_id'].nunique()
mimic_patients = df_icustays['subject_id'].nunique()
eicu_stays = df_patient['patientunitstayid'].nunique()
eicu_patients = df_patient['uniquepid'].nunique()

print(f"MIMIC-IV  — ICU stays: {mimic_stays:,} | Patients: {mimic_patients:,}")
print(f"eICU      — ICU stays: {eicu_stays:,} | Patients: {eicu_patients:,}")
print(f"COMBINED  — ICU stays: {mimic_stays + eicu_stays:,} | Patients: {mimic_patients + eicu_patients:,}")

print("\n" + "=" * 50)
print("LENGTH OF STAY COMPARISON")
print("=" * 50)
print(f"MIMIC-IV  — Average stay: {df_icustays['los'].mean():.2f} days")
print(f"eICU      — Average stay: {df_patient['los_hours'].mean():.2f} hours ({df_patient['los_hours'].mean()/24:.2f} days)")

print("\n" + "=" * 50)
print("VITAL SIGNS FILE SCALE")
print("=" * 50)
print(f"vitalPeriodic  — {len(df_periodic_full):,} rows (Philips direct monitor output)")
print(f"vitalAperiodic — {len(df_vital_full):,} rows")
print(f"Total eICU vital signs rows: {len(df_periodic_full) + len(df_vital_full):,}")
```

```python
# Monitor activity signal quality summary
print("=" * 50)
print("MONITOR ACTIVITY SIGNAL QUALITY")
print("=" * 50)

total_periodic = len(df_periodic_full)
hr_present = total_periodic - df_periodic_full['heartrate'].isnull().sum()
sao2_present = total_periodic - df_periodic_full['sao2'].isnull().sum()
resp_present = total_periodic - df_periodic_full['respiration'].isnull().sum()

print(f"vitalPeriodic total rows: {total_periodic:,}")
print(f"heartrate present:   {hr_present:,} ({hr_present/total_periodic*100:.1f}%)")
print(f"sao2 present:        {sao2_present:,} ({sao2_present/total_periodic*100:.1f}%)")
print(f"respiration present: {resp_present:,} ({resp_present/total_periodic*100:.1f}%)")

total_aperiodic = len(df_vital_full)
nibp_present = total_aperiodic - df_vital_full['noninvasivesystolic'].isnull().sum()
print(f"\nvitalAperiodic total rows: {total_aperiodic:,}")
print(f"noninvasivesystolic present: {nibp_present:,} ({nibp_present/total_aperiodic*100:.1f}%)")
```
